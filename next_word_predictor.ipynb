```python
# Next Word Prediction using LSTM (Minimal)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
# nltk.download('wordnet')
# nltk.download('punkt')
# nltk.download('punkt_tab')
```

```python
# 1. Loading dataset
paragraph = open('dune_Messiah.txt', 'r', encoding='utf-8').read()
sentences = nltk.sent_tokenize(paragraph)
```

```python
corpus = ' '.join(sentences)
corpus = re.sub('[^a-zA-Z0-9]', ' ', corpus).lower()
corpus = ' '.join(corpus.split())
```

```python
corpus
```

```python
text = "i love deep learning and i love nlp"
```

```python
# 2. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts([corpus])
seq = tokenizer.texts_to_sequences([corpus])[0]
vocab_size = len(tokenizer.word_index) + 1
```

```python
# 3. Create input-output sequences
X, y = [], []
for i in range(1, len(seq)):
    X.append(seq[:i])
    y.append(seq[i])

X = pad_sequences(X)
y = tf.keras.utils.to_categorical(y, vocab_size)
```

```python
# 4. Model
model = Sequential([
    Embedding(vocab_size, 10),
    LSTM(64),
    Dense(vocab_size, activation="softmax")
])

model.compile(loss="categorical_crossentropy", optimizer="adam")
```

```python
# 5. Predict next word
def predict_next(text, n=1):
    for _ in range(n):
        s = tokenizer.texts_to_sequences([text])[0]
        s = pad_sequences([s], maxlen=X.shape[1])
        word_id = model.predict(s, verbose=0).argmax()
        text += " " + tokenizer.index_word[word_id]
    return text
```

```python
text = "dune messiah is the most"
for i in range(20):
    output = predict_next(text)
    text = output
```
