```python
import tensorflow as tf
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.layers import LSTM, Dense
```

```python
# Vocabulary size
voc_size = 5000
```

### One-Hot Encoding

```python
messages = X.copy()
```

```python
messages.head()
```

```python
messages = messages.iloc[:10000, :]
```

```python
messages.shape
```

```python
# messages.reset_index(inplace=True)
```

```python
import nltk
import re
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
corpus = []
for title in messages['title']:
    # re.sub -> (substitute) all the special characters with space
    review = re.sub('[^a-zA-Z]', ' ', title)
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus[:5]
```

```python
onehot_repr = [one_hot(words, voc_size) for words in corpus]
```

```python
onehot_repr
```

```python
corpus[1]
```

```python
onehot_repr[1]
```

### Embedding Representation

```python
sent_length = 20
embedding_docs = pad_sequences(onehot_repr, padding='pre', maxlen=sent_length)
```

```python
embedding_docs
```

```python
# Creating model
embedding_vector_features = 40 # features representation
model = Sequential()
model.add(Embedding(voc_size, embedding_vector_features, input_length=sent_length))
model.build(input_shape=(None, sent_length)) 
model.add(LSTM(100)) # LSTM layer with 100 neurons
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
len(embedding_docs), y.shape
```

```python
import numpy as np
X_final = np.array(embedding_docs)
y_final = np.array(y)
```

```python
X_final.shape, y_final.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_final, y_final, test_size=0.2, random_state=42)
```

```python
# Model Training
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10, batch_size=64)
```

### Performance Metrics and Accuracy

```python
y_pred = model.predict(X_test)
```

```python
y_pred = np.where(y_pred > 0.5, 1, 0) # AUC ROC Curve
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
