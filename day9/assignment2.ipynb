### Steps
- step 1: Data loading and preprocessing
- step 2: One-Hot Encoding
- Step 3: Padding
- Step 4: Embedding

```python
# step 1: Data loading and preprocessing
# dataset: https://www.kaggle.com/datasets/lakshmi25npathi/imdb-dataset-of-50k-movie-reviews
import pandas as pd
dataset = pd.read_csv('IMDB Dataset.csv')
```

```python
dataset.head()
```

```python
dataset = dataset.iloc[:, 0].values
```

```python
dataset.shape
```

```python
# step 2: One-Hot Encoding
from tensorflow.keras.preprocessing.text import one_hot
```

```python
voc_size = 5000
encoded_dataset = [one_hot(review, voc_size) for review in dataset]
```

```python
max = 0
x = 0
for i in encoded_dataset:
    if len(i) > max:
        max = len(i)
        x = i
print(max)
```

```python
# Step 3: Padding
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length = 500
embedded_dataset = pad_sequences(encoded_dataset, padding='pre', maxlen=sent_length)
print(embedded_dataset.shape)
```

```python
# Step 4: Embedding
dim = 10
model = Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_length))
model.compile('adam', 'mse')
model.build(input_shape=(None, sent_length)) # Explicitly build the model
```

```python
model.summary()
```

```python
model.predict(np.expand_dims(embedded_dataset[0], axis=0))
```

```python
# Predicting on the entire dataset can cause OOM; predict on a single batch/sample instead
model.predict(embedded_dataset)
```
